```python
import sqlite3
```

```python
conn = sqlite3.connect('ecommerce.db')
cursor = conn.cursor()
```

```python
try:
    cursor.execute('''
        Create Table Customers (
        	customer_id integer primary key autoincrement,
        	customer_name text not null,
        	email text not null unique,
        	phone text not null unique,
        	address text not null,
        	customer_type text not null CHECK (customer_type  in (
        											'Regular', 'Non-Regular', 'VIP'
        										))
        );
    ''')
    print('Customer table created successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
```

```python
try:
    cursor.execute('''
        Create Table Staffs (
            staff_id integer primary key autoincrement,
            staff_name text not null,
            staff_type text not null check (staff_type in ('Full-Time', 'Part-Time', 'Contract')),
            staff_position text not null check (staff_position in ('Intern', 'Junior', 'Senior')),
            staff_email text not null unique
        )
    ''')
    print('Staff table created successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')

```

```python
try:
    cursor.execute('''
        Create Table Orders (
            order_id integer primary key autoincrement,
            order_status text not null check (order_status in ('Completed', 'Processing', 'Pending', 'Rejected')),
            -- YYYY-MM-DD
            order_date date not null,
            order_desc text,
            customer_id integer not null,
            staff_id integer not null,
            foreign key (customer_id) references Customers(customer_id),
            foreign key (staff_id) references Staffs(staff_id)
        );
    ''')
    print('Staff table created successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
```

```python
try:
    cursor.execute('''
        Insert into Customers (customer_name, email, phone, address, customer_type)
        values 
        ('Sita', 'sita@gmail.com', '5432167890', 'Lalitpur', 'Regular'),
        ('Bob', 'bob@gmail.com', '1230987654', 'Bhaktapur', 'Regular'),
        ('Arthur', 'arthur@gmail.com', '0987123456', 'Makwanpur', 'Non-Regular');
    ''')
    conn.commit()
    print('Customer data insert successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
except sqlite3.IntegrityError as ie:
    print(f'Error: {ie}')
```

###Like

```python
'''
wildcards
-----------
% -> pattern matching
_ -> single word
'''
```

```python
# Find customer name whose name start with letter 'S'
```

```python
try:
    cursor.execute('''
    select * from customers where customer_name like 'S%';
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
    select * from customers where customer_name like 'm%';
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```
