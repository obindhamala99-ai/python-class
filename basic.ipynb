```python
# if - elif-else
#Temperature
temperature = float(input('Enter temperature value:'))
if temperature >= 28:
    print('Its hot today')
elif temperature >= 15 and temperature < 28:
    print('Its warm today')
else:
    print('Its cold today')
```

```python
'''
-80 - 100 -> grade:A
-60 - 79 -> grade:B
-50 - 59 -> grade:C
-40 - 49 -> grade:D
-> Fail
'''
grade = float (input('Enter grade value:'))
if grade >= 80 and grade <=100:
    print('Grade: A')
elif grade >= 60 and grade <=79:
    print ('Grade B')
elif grade >= 50 and grade <=59:
    print ('Grade C')
elif grade >= 40 and grade <= 49:
    print ('Grade D')
else:
    print('Its fail')
```
